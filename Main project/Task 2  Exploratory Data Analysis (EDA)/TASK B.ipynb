```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,5))
df = pd.read_excel("SuperStoreOrders_cleaned.xlsx")

#B. TIME BASED ANALYSIS
#MONTHLY SALES TREND
df['order_date'] = pd.to_datetime(df['order_date'])
df['month'] = df['order_date'].dt.month_name()
df['year'] = df['order_date'].dt.year
df[['order_date','month','year']].head()
month_order = ['January','February','March','April','May','June',
               'July','August','September','October','November',
               'December']
df['month'] = pd.Categorical(
    df['month'],
    categories=month_order,
    ordered=True
)
monthly_sales = df.groupby('month')['sales'].sum()
monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print("INSIGHTS:Sales increase during festive and holiday seasons.")
print("RECOMMENDATION:Maintain higher inventory during peak sales months.")

#YEARLY PROFIT TREND
yearly_profit = df.groupby('year')['profit'].sum()
yearly_profit.plot(kind='line', marker='o')
plt.title("Yearly Profit Trend")
plt.ylabel("Profit")

plt.show()
print("INSIGHTS:Profit shows growth trend over the years.")
print("RECOMMENDATION:Continue successful strategies that improved profitability.")

#SEASONAL SALES PATTERN
sns.boxplot(x='month', y='sales', data=df)
plt.xticks(rotation=90)
plt.title("Seasonal Sales Pattern")

plt.show()
print("INSIGHTS:Some months experience unusually high sales activity.")
print("RECOMMENDATION:Plan seasonal promotions during high-demand periods.")
```
