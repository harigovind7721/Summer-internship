```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,5))
df = pd.read_excel("SuperStoreOrders_cleaned.xlsx")

#TOP 10 PROFITABLE PRODUCTS
top_products = df.groupby('product_name')['profit'].sum() \
                 .sort_values(ascending=False).head(10)
top_products = top_products.sort_values()

plt.figure(figsize=(10,6))

top_products.plot(kind='barh')
plt.title("Top 10 Profitable Products")
plt.xlabel("Profit")
plt.show()
print("INSIGHTS:A small group of products contributes major profits.")
print("RECOMMENDATION:Promote high-performing products through advertisements.")

#BOTTOM 10 LOSS-MAKING PRODUCTS
bottom_products = df.groupby('product_name')['profit'].sum() \
                    .sort_values().head(10)
bottom_products.plot(kind='barh')
plt.title("Bottom 10 Loss-Making Products")
plt.xlabel("Loss")
plt.show()
print("INSIGHTS:Some products continuously reduce company profitability.")
print("RECOMMENDATION:Reduce discounts or discontinue low-performing products.")
```
